```python
import json
import re

with open("data/raw/resumes/train.json", "r", encoding="utf-8", errors="surrogatepass") as f:
    raw_text = f.read()

# Remove unpaired/lone surrogate characters that break JSON parsing
cleaned_text = re.sub(r'[\ud800-\udfff]', '', raw_text)

data = json.loads(cleaned_text)
resume_train_df = pd.DataFrame(data)
print(resume_train_df.shape)
print(resume_train_df.columns)
resume_train_df.head()
```

```python
subset_cols = ['question', 'category', 'role', 'experience', 'difficulty', 'source_type', 'ideal_answer']
hr_df = hr_df.drop_duplicates(subset=subset_cols).reset_index(drop=True)
print(hr_df.shape)  # should now be ~34,560 rows
```

```python
# Standardize hr_df
hr_clean = hr_df.rename(columns={
    "ideal_answer": "answer"
})[["question", "answer", "category", "role", "difficulty"]].copy()
hr_clean["domain"] = "HR"

# Standardize tech_df
tech_clean = tech_df.rename(columns={
    "Question": "question",
    "Answer": "answer",
    "Category": "category",
    "Difficulty": "difficulty"
})[["question", "answer", "category", "difficulty"]].copy()
tech_clean["role"] = "Software Engineer"  # tech_df has no role column, default assumption
tech_clean["domain"] = "Technical"

# Combine into one unified question bank
unified_df = pd.concat([hr_clean, tech_clean], ignore_index=True)
print(unified_df.shape)
unified_df.sample(5)
```
