```python
resume_core = resume_df[[
    'career_objective', 'skills', 'degree_names', 'major_field_of_studies',
    'professional_company_names', 'positions', 'responsibilities',
    'job_position_name', 'educationaL_requirements', 'experiencere_requirement',
    'skills_required', 'matched_score'
]].copy()

resume_core = resume_core.dropna(subset=['skills'])
print(resume_core.shape)
resume_core.sample(3)
```

```python
def extract_entities(row):
    text = row['text']
    entities = []
    for start, end, label in row['annotations']:
        entities.append({'entity_text': text[start:end], 'label': label})
    return entities

resume_train_df['entities'] = resume_train_df.apply(extract_entities, axis=1)
print(resume_train_df.iloc[0]['entities'][:5])
```

```python
all_labels = set()
for entities in resume_train_df['entities']:
    for e in entities:
        all_labels.add(e['label'])
print(all_labels)
```

```python
def extract_skills(entities):
    return list(set(e['entity_text'] for e in entities if e['label'] == 'SKILL'))

resume_train_df['extracted_skills'] = resume_train_df['entities'].apply(extract_skills)
print(resume_train_df[['text', 'extracted_skills']].iloc[0])
```

```python
resume_categorized = resume_jd_df[['ID', 'Resume_str', 'Category']].copy()
resume_categorized['Resume_str'] = resume_categorized['Resume_str'].str.strip()

print(resume_categorized.shape)
print(resume_categorized['Category'].nunique())
print(resume_categorized['Category'].unique())
resume_categorized.sample(2)
```

```python
import re

# --- Source 1: resume_df (structured) ---
def combine_text_1(row):
    parts = [str(row.get('career_objective', '')), str(row.get('responsibilities', ''))]
    return " ".join(p for p in parts if p != 'nan')

resume_1 = pd.DataFrame({
    'resume_id': resume_core.index.astype(str) + "_r1",
    'source': 'resume_df',
    'raw_text': resume_core.apply(combine_text_1, axis=1),
    'skills': resume_core['skills'],
    'category': resume_core['job_position_name']
})

# --- Source 2: resume_train_df (NER-extracted) ---
# Clean up skills — split on bullets/commas since raw spans dumped whole lines
def clean_skill_list(skills):
    cleaned = []
    for s in skills:
        parts = re.split(r'[•,]', s)
        cleaned.extend([p.strip() for p in parts if p.strip() and len(p.strip()) < 50])
    return list(set(cleaned))

resume_train_df['clean_skills'] = resume_train_df['extracted_skills'].apply(clean_skill_list)

resume_2 = pd.DataFrame({
    'resume_id': resume_train_df.index.astype(str) + "_r2",
    'source': 'resume_train_df',
    'raw_text': resume_train_df['text'],
    'skills': resume_train_df['clean_skills'],
    'category': None
})

# --- Source 3: resume_jd_df (categorized plain text) ---
resume_3 = pd.DataFrame({
    'resume_id': resume_categorized['ID'].astype(str) + "_r3",
    'source': 'resume_jd_df',
    'raw_text': resume_categorized['Resume_str'],
    'skills': None,
    'category': resume_categorized['Category']
})

# --- Merge all three ---
unified_resumes = pd.concat([resume_1, resume_2, resume_3], ignore_index=True)
print(unified_resumes.shape)
unified_resumes.sample(5)
```

```python
import re

def remove_surrogates(x):
    if isinstance(x, list):
        return [re.sub(r'[\ud800-\udfff]', '', str(item)) for item in x]
    elif isinstance(x, str):
        return re.sub(r'[\ud800-\udfff]', '', x)
    else:
        return x

for col in ['raw_text', 'skills', 'category']:
    unified_resumes[col] = unified_resumes[col].apply(remove_surrogates)

unified_resumes.to_csv("data/processed/text/unified_resumes.csv", index=False)
unified_job_descriptions.to_csv("data/processed/text/unified_job_descriptions.csv", index=False)

print("All unified datasets saved to data/processed/text/")
```

```python
import re

def clean_text(text):
    if not isinstance(text, str):
        return text
    text = text.lower()
    text = re.sub(r'<.*?>', ' ', text)          # remove HTML tags
    text = re.sub(r'http\S+|www\S+', ' ', text) # remove URLs
    text = re.sub(r'[^a-z0-9\s.,!?\'-]', ' ', text)  # remove special chars, keep basic punctuation
    text = re.sub(r'\s+', ' ', text).strip()    # collapse extra whitespace
    return text
```

```python
from nltk.stem import WordNetLemmatizer
from nltk.tokenize import word_tokenize

lemmatizer = WordNetLemmatizer()

def tokenize_lemmatize(text):
    if not isinstance(text, str) or text.strip() == '':
        return []
    tokens = word_tokenize(text)
    lemmatized = [lemmatizer.lemmatize(tok) for tok in tokens]
    return lemmatized
```

```python
unified_df['question_tokens'] = unified_df['question_clean'].apply(tokenize_lemmatize)
unified_df['answer_tokens'] = unified_df['answer_clean'].apply(tokenize_lemmatize)
print(unified_df[['question_clean', 'question_tokens']].sample(2))
```

```python
from nltk.corpus import stopwords
nltk.download('stopwords')

stop_words = set(stopwords.words('english'))

def remove_stopwords(tokens):
    return [tok for tok in tokens if tok.lower() not in stop_words and tok.isalpha()]
```

```python
unified_df['question_keywords'] = unified_df['question_tokens'].apply(remove_stopwords)
unified_resumes['raw_text_keywords'] = unified_resumes['raw_text_tokens'].apply(remove_stopwords)
unified_job_descriptions['description_keywords'] = unified_job_descriptions['description_tokens'].apply(remove_stopwords)

print(unified_df[['question_tokens', 'question_keywords']].sample(2))
```

```python
category_counts = unified_df['category'].value_counts()
print(category_counts)
print("\nCategories with only 1 row:", (category_counts == 1).sum())
```

```python
category_counts = unified_df['category'].value_counts()
rare_categories = category_counts[category_counts < 10].index.tolist()

unified_df['category_grouped'] = unified_df['category'].apply(
    lambda x: 'Other' if x in rare_categories else x
)

print(unified_df['category_grouped'].value_counts())
```

```python
from sklearn.model_selection import train_test_split

train_df, temp_df = train_test_split(
    unified_df, test_size=0.30, stratify=unified_df['category_grouped'], random_state=42
)

val_df, test_df = train_test_split(
    temp_df, test_size=0.50, stratify=temp_df['category_grouped'], random_state=42
)

print("Train:", train_df.shape)
print("Val:", val_df.shape)
print("Test:", test_df.shape)
```
